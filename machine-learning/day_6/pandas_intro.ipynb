```python
data = {
    "Product": ['Phone', 'Laptop', 'Tablet', 'Watch'],
    "price": [50000, 70000, 30000, 15000],
    "Quantity": [5, 3, 7, 10]
}
df = pd.DataFrame(data)
print(df)
```

```python
df['TotalValue'] = df['price'] * df['Quantity']
print(df)
```

```python
average_price = df['price'].mean()
print(average_price)
```
