```python
import pandas as pd
df = pd.read_csv("StudentsPerformance.csv")
print(df.head(7))
```

```python
print(df.columns)
```

```python
print(df.shape)
print()
print(df.info())
print()
print(df.describe())
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Check for duplicate rows
print(df.duplicated().sum())
```

```python
# Remove duplicates if any
df.drop_duplicates(inplace=True)
```

```python
#Summary Functions
print("Average of Math score:",df["math score"].mean())
print("Maximum reading score:",df["reading score"].max())
print(df['gender'].unique())
print(df["math score"].value_counts())
```

```python
#Quick Mini Data Exploration
print(df["gender"].value_counts())
print()
print("Average Marks by Gender:")
print(df.groupby('gender')['math score'].mean())
```

```python
common_parental_level_of_education = df["parental level of education"].mode()[0]
print(common_parental_level_of_education)
```
