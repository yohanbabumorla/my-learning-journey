```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ['Yohan','Ravi','Anu','Teja','Kiran'],
    "Age": [20, 21, 19, 22, 23],
    'Marks': [85, 90, 75, 60, 95]
    
})
print(df)
```

```python
#select column
print(df['Name'])
#select multiple columns
print("\n",df[["Name","Age"]])
```

```python
print(df.iloc[0])   # First row by index position
print(df.loc[2])    # Row by label (index value)
```

```python
print(df[1:4])   # Rows from index 1 to 3
```

```python
"""Conditional Filtering"""
```

```python
# Students with marks > 80
higher_rankers = df[df["Marks"] >80]
print(higher_rankers)
```

```python
# Students younger than 22 and marks > 70
filtered = df[(df["Age"] < 22) & (df["Marks"] > 70)]
print(filtered)
```

```python
"Handling Missing Values (NaN)"
```

```python
data = {
    'Name': ['Yohan', 'Ravi', 'Anu', 'Teja'],
    'Age': [20, None, 19, 22],
    'Marks': [85, 90, None, 60]
}

df = pd.DataFrame(data)
print(df)
```

```python
"Fill missing values"
```

```python
# print(df.isnull())
```

```python
# print(df.isnull().sum())
```

```python
df.fillna({'Age': df['Age'].mean()},inplace=True)
print(df)
```

```python
df.fillna({'Marks': 0}, inplace=True)
print(df)
```

```python
df.rename(columns={'Marks': 'Score'}, inplace=True) #Rename column
df.drop('Age', axis=1, inplace=True)  # Drop a column
print(df)
```

```python
df.sort_values(by='Score', ascending=False) #sort by score
```

```python
df.sort_values(by=['Score', 'Name'])
```

```python
data = {
    'Name': ['Ram', 'Krish', 'Latha', 'Hari', 'Mani'],
    'Age': [23, None, 21, 25, 22],
    'Department': ['IT', 'HR', 'Finance', None, 'IT'],
    'Salary': [50000, 45000, None, 52000, 48000]
}
df =pd.DataFrame(data)

print(df)
```

```python
# Tasks:
# 1. Display names of employees from the 'IT' department
# 2. Fill missing Salary with average salary
# 3. Drop rows with missing Department
# 4. Rename 'Department' → 'Dept'
# 5. Sort employees by Salary in descending order
```

```python
emp_it = df.loc[df['Department'] == 'IT']
print(emp_it['Name'])
```

```python
df.fillna({'Salary': df['Salary'].mean()}, inplace=True)
print(df)
```

```python
df.dropna(subset=['Department'])
```

```python
df.rename(columns={'Department': 'Dept'}, inplace=True)
print(df)
```

```python
df.sort_values(by="Salary", ascending=False)
```
