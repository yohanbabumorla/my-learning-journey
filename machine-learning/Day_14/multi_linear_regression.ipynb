### Multiple Linear Regression

```python
import pandas as pd

data = {
    'Experience': [1,2,3,4,5,6,7,8],
    'SkillScore': [40,50,60,65,70,75,80,85],
    'Certifications': [0,1,1,2,2,3,3,4],
    'Salary': [30000,35000,40000,45000,50000,55000,60000,65000]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.model_selection import train_test_split

X = df[['Experience','SkillScore','Certifications']] #multiple features
y = df['Salary'] #target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print("Coefficients (Weights):", model.coef_)
print("Intercept (Bias):", model.intercept_)
```

```python
predictions = model.predict(X_test)
print("Predicted salaries:",predictions)
print("Actual Salaries:", list(y_test))
```

```python
from sklearn.metrics import r2_score

score = r2_score(y_test, predictions)
print('R2 Score:', score)
```

```python
import matplotlib.pyplot as plt

plt.scatter(y_test, predictions, color='purple')
plt.xlabel('Actual Salary')
plt.ylabel('Predicted Salary')
plt.title('Actual Vs Predicted Salary')

plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red')
plt.show()
```
