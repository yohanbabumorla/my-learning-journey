```python
# Create a dataset of 50 random exam scores between 0–100.

# Plot a histogram to show the score distribution.

# Plot a bar chart showing average marks in 3 subjects.

# Create a pie chart showing time spent on daily activities (Sleep, Study, Games, Internet, Exercise).

# Display all 3 charts as subplots in one figure.
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = np.random.randint(0, 100, size=50)
plt.hist(scores,bins=7,color='green',edgecolor='black')
plt.title("Scores Distribution of an Exam")
plt.xlabel("Scores")
plt.ylabel("Students")
plt.yticks([0,2,4,6,8,10])
plt.show()
```

```python
math_scores = np.random.randint(0,100,50)
science_scores = np.random.randint(0,100,size=50)
english_scores = np.random.randint(0,100,size=50)
avrg_math = np.mean(math_scores)
avrg_science = np.mean(math_scores)
avrg_english = np.mean(english_scores)
average = [avrg_math,avrg_science,avrg_english]
subjects = ["Maths","Science","English"]
plt.bar(subjects,average, color="teal")
plt.title("Average marks per subjects")
plt.xlabel('subjects')
plt.ylabel("average scores")
plt.show()
```

```python
activities = ["Sleep","Study","Games","Internet","Exercise"]
hours = [8,8,1,6,1]
plt.pie(hours,labels=activities,autopct='%1.1f%%',startangle=140)
plt.title("Hours spent by Daily activities")
plt.show()
```

### Display all 3 charts as Subplots in one figure

```python
fig,axs = plt.subplots(1,3, figsize=(15,4))

axs[0].hist(scores,bins=7,color='green',edgecolor='black')
axs[0].set_title("Scores Distribution of an Exam")

axs[1].bar(subjects,average, color="teal")
axs[1].set_title("Average marks from each subject")

axs[2].pie(hours,labels=activities,autopct='%1.1f%%',startangle=140)
axs[2].set_title("Time spent on daily activities")

plt.suptitle("The class Student's dashboard")
plt.show()
```
