```python
# 1️⃣ Subplots – Multiple Graphs in One Figure

# Subplots let you show multiple related visualizations side by side — perfect for comparisons
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [5,15,25,35,45]

# Create 1 row, 2 columns of plots
fig, axs = plt.subplots(1,2,figsize=(10,4))

#First plot
axs[0].plot(x, y1, color='blue')
axs[0].set_title("Dataset 1")

#Second plot
axs[1].plot(x,y2,color='red')
axs[1].set_title("Dataset 2")

plt.suptitle("Comparison using Subplots")
plt.show()
```

```python
# #📊 2️⃣ Histogram – Understanding Data Distribution
# Histograms are used to see how data is distributed — they’re perfect for understanding patterns like exam marks, ages, or incomes.
```

```python
import numpy as np 
import matplotlib.pyplot as plt

ages = np.random.randint(18,30,100)

plt.hist(ages, bins=6, color='green', edgecolor='black')
plt.title("Age Distribution of Students")
plt.xlabel("Age Groups")
plt.ylabel("Frequency")
plt.show()
```

```python
# 🥧 3️⃣ Pie Charts – Visualizing Percentages

# Pie charts are great for showing categorical data proportions.
```

```python
labels = ['Python','Java','C++','JavaScript']
students = [50,25,15,10]

plt.pie(students, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Programming Language Popularity')
plt.show()
```

```python
# 4️⃣ Combining Charts in Subplots

# Let’s make a simple dashboard 👇
```

```python
import numpy as np
import matplotlib.pyplot as plt

sales = [200, 250, 180, 300, 280, 350]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

profits = [20, 30, 15, 40, 35, 45]
categories = ['Phones', 'Laptops', 'Watches', 'Accessories']
shares = [40, 30, 20, 10]

fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# Line plot
axs[0].plot(months, sales, color='blue', marker='o')
axs[0].set_title("Monthly Sales")

# Bar chart
axs[1].bar(months, profits, color='orange')
axs[1].set_title("Monthly Profits")

# Pie chart
axs[2].pie(shares, labels=categories, autopct='%1.1f%%')
axs[2].set_title("Product Share")

plt.suptitle("Business Overview Dashboard")
plt.show()

# ✅ You’ve just made your first mini dashboard!
# This is what data analysts create to summarize business or dataset insights.
```
