```python
import matplotlib.pyplot as plt
```

```python
# Create a dataset of monthly sales:

# months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
# sales = [20000, 25000, 18000, 30000, 28000, 35000]


# Plot a line chart of sales over months.

# Add a bar chart for the same data.

# Add a title, axis labels, and grid lines.

# Change line color and style for better visuals.
```

```python
months = ['Jan','Feb','Mar','Apr','May','Jun']
sales = [20000, 25000, 18000, 30000, 28000, 35000]

plt.plot(months,sales)
plt.title("Months and Sales")
plt.xlabel("months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.bar(months,sales, color='teal')
plt.title('Months Sales')
plt.xlabel("Months",fontsize=12)
plt.ylabel("Sales",fontsize=12)
plt.show()
```

```python
plt.plot(months,sales, color="blue", linestyle='--',marker="*", linewidth=2)
plt.title("Months Sales Graph", fontsize=15,color="orange")
plt.xlabel("Months", fontsize=15, color="teal")
plt.ylabel("Sales", fontsize=15, color="teal")
plt.grid(True)
plt.show()
```
