```python
import matplotlib.pyplot as plt
```

```python
#creating a simple line plot
x = [1,2,3,4,5]
y = [10,20,25,30,50]
plt.plot(x,y)
plt.title("Simple line plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
#Bar Charts
students = ['Yohan','Ravi','Anu','Teja']
marks = [85,90,75,60]
plt.bar(students,marks,color = 'teal')
plt.title("Students Marks comparision")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#scatter plots
age = [18, 19, 20, 21, 22, 23]
marks = [75, 80, 70, 85, 60, 90]

plt.scatter(age,marks,color='purple')
plt.title("Age Vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.show()
```

```python
#Customizing Plots
x= [1,2,3,4,5]
y = [2,3,5,7,11]

plt.plot(x, y, color='red', linestyle="--", marker= "o", linewidth=2)
plt.title("Customized Line Graph", fontsize=14, color='blue')
plt.xlabel("X Values", fontsize=12)
plt.ylabel("Y Values", fontsize=12)
plt.grid(True)
plt.show()
```

```python
#multiple plot in one graph
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [5,15,25,35,45]

plt.plot(x,y1 , label='Dataset1',color='blue')
plt.plot(x,y2, label='Dataset 2', color='orange')
plt.title("Multiple line plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```
