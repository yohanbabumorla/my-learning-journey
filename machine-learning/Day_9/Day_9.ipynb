```python
#Day 9 – Data Cleaning, Transformation & Encoding
"""Handle missing and duplicate data

Transform numerical and categorical features

Encode text values into numbers (for ML models)"""
```

```python
import pandas as pd
import numpy as np
data = {
    "Name": ['Yohan', 'Ravi', 'Anu', 'Teja', 'Kiran'],
    "Age": [20, np.nan, 19, 22, np.nan],
    'Marks': [85, 90, np.nan, 60, 95],
    'City': ['Hyd', 'Vizag', np.nan, 'Hyd', 'Chennai']
}
df = pd.DataFrame(data)
print(df)
```

```python
#🧠 1️⃣ Handling Missing Data (NaN)
print(df.isnull().sum())
```

```python
df.fillna({'Age': df["Age"].mean()}, inplace=True)
df.fillna({"Marks": df['Marks'].median()}, inplace=True)
df.fillna({"City": df["City"].mode()[0]}, inplace=True)
print(df)
```

```python
#🧱 2️⃣ Handling Duplicates
# Add duplicate rows for demo
df.loc[len(df)] = ['Ravi', 21, 90, 'Vizag']

print("Duplicates Count:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

```python
# #⚙️ 3️⃣ Data Transformation
# Transformations make numeric data consistent and ML-friendly.
```

```python
#Convert Data Types
df['Age'] = df['Age'].astype(int)
print(df.dtypes)
```

```python
# Scaling & Normalization

# Normalization helps when features have very different ranges (e.g., age in 20s, marks in 100s).
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Age", "Marks"]] = scaler.fit_transform(df[['Age', 'Marks']])
print(df)
```

```python
df['Performance'] = ['Good' if m >= 0.75 else 'Average' for m in df['Marks']]
print(df[['Name', 'Marks', 'Performance']])
```
