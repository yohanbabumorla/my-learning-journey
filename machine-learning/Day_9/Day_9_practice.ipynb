```python
import pandas as pd
import numpy as np


data = pd.read_csv("cars.csv")
df = pd.DataFrame(data)
print(df.head(7))
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate data", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df.duplicated().sum())
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

df["fuel_encoded"] = encoder.fit_transform(df["fuel"])

print(df.head(7))
```

```python
print(df["fuel_encoded"].value_counts())
```

```python
print(df["fuel"].value_counts())
```

```python
df_encoded = pd.get_dummies(df, columns = ["fuel"])
print(df_encoded.head(7))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['km_driven', 'selling_price']] = scaler.fit_transform(df[['km_driven', 'selling_price']])
print(df.head(7))
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
