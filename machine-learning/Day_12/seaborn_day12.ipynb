```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load a sample dataset
tips = sns.load_dataset("tips")
print(tips.head())
```

```python

sns.lineplot(x='total_bill', y='tip', data=tips)
plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.title("Tips by Gender")
plt.show()
```

```python
sns.barplot(x='day', y='total_bill',hue= "day", data=tips, palette='coolwarm')
plt.title("Average Bill per Day")
plt.show()
```

```python
sns.histplot(tips['total_bill'], kde=True, color='orange')
plt.title("Distribution of Total Bills")
plt.show()
```

```python
sns.boxplot(x="day", y='total_bill', data=tips)
plt.title("Bill Amount Spread by Day")
plt.show()
```

```python
sns.pairplot(tips, hue="sex")
plt.suptitle("Pairwise Relationships", y=1.02)
plt.show()
```

```python
# Select only numeric columns before computing correlation
corr = tips.select_dtypes(include='number').corr()

# Plot the heatmap
sns.heatmap(corr, annot=True, cmap
            
            ='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
